```python
import pandas as pd
```

```python
df1 = pd.read_excel('clientes_comp.xlsx', sheet_name='Clientes1') 
df1
```

```python
# Resumo dos dados: tipos de variáveis, nulos e valores distintos
resumo_dados = pd.DataFrame({
    "tipo_variavel": df1.dtypes,
    "nulos": df1.isna().sum(),
    "valores_distintos": df1.nunique(dropna=True)
}).reset_index().rename(columns={"index": "coluna"})

resumo_dados
```

```python
quantidade_por_categoria = (
    df1.groupby("Categoria do Produto")["QTD"]
    .sum()
    .sort_values(ascending=False)
    .rename("Quantidade Vendida")
)

quantidade_por_categoria
```

```python
maior_venda_por_categoria = (
    df1.groupby(["Categoria do Produto", "Produto"])["QTD"]
    .sum()
    .rename("Quantidade Vendida")
    .reset_index()
)

maiores_produtos = maior_venda_por_categoria[
    maior_venda_por_categoria["Quantidade Vendida"]
    == maior_venda_por_categoria.groupby("Categoria do Produto")["Quantidade Vendida"]
    .transform("max")
]

display(quantidade_por_categoria.rename("Quantidade Vendida").to_frame())
display(maiores_produtos)
```

```python
vendas_por_categoria = (
    df1.groupby("Categoria do Produto")["Venda Total"]
    .sum()
    .sort_values(ascending=False)
    .rename("Total de Vendas")
)

display(vendas_por_categoria.to_frame())

valor_total_vendas = df1["Venda Total"].sum()
print(f"Valor total em vendas: R$ {valor_total_vendas:,.2f}")
```

```python
# Valor arrecadado por categoria de produtos
vendas_por_categoria = (
    df1.groupby("Categoria do Produto")["Venda Total"]
    .sum()
    .sort_values(ascending=False)
    .rename("Total de Vendas")
)

display(vendas_por_categoria.to_frame())

for categoria, total in vendas_por_categoria.items():
    print(f"{categoria}: R$ {total:,.2f}")
```

```python
produtos_ordenados = sorted(df1["Produto"].dropna().unique())
produtos_ordenados
```

```python
# Quantidade vendida e número de registros por forma de pagamento
vendas_por_pagamento = (
    df1.groupby("Forma de Pagamento")
    .agg(
        Quantidade_Vendida=("QTD", "sum"),
        Numero_de_Vendas=("QTD", "size")
    )
    .sort_values("Quantidade_Vendida", ascending=False)
)

# Total em vendas por produto
total_por_produto = (
    df1.groupby("Produto")["Venda Total"]
    .sum()
    .sort_values(ascending=False)
    .rename("Total de Vendas (R$)")
)

display(vendas_por_pagamento)
display(total_por_produto.to_frame())
```

```python
# Quantidade vendida por produto
quantidade_por_produto = (
    df1.groupby("Produto")["QTD"]
    .sum()
    .sort_values(ascending=False)
    .rename("Quantidade Vendida")
)

# Soma das quantidades vendidas por categoria
display(quantidade_por_produto.to_frame())
display(quantidade_por_categoria.rename("Quantidade Vendida").to_frame())
```

```python
# Quantidade total vendida de cada produto em cada categoria
tabela_quantidade_por_categoria = (
    df1.pivot_table(
        index="Categoria do Produto",
        columns="Produto",
        values="QTD",
        aggfunc="sum",
        fill_value=0
    )
)

display(tabela_quantidade_por_categoria)
```

```python
# Quantidade de produtos vendidos por dia
vendas_por_dia = (
    df1.groupby("Data de Publicação", as_index=False)["QTD"]
    .sum()
    .rename(columns={"QTD": "Quantidade Vendida"})
    .sort_values("Data de Publicação")
)

display(vendas_por_dia.head(10))

# Vendas diárias por categoria
vendas_por_dia_categoria = (
    df1.groupby(["Data de Publicação", "Categoria do Produto"], as_index=False)["QTD"]
    .sum()
    .rename(columns={"QTD": "Quantidade Vendida"})
    .sort_values(["Data de Publicação", "Quantidade Vendida"], ascending=[True, False])
)

display(vendas_por_dia_categoria.head(20))

# Categoria que mais vendeu em cada dia
categoria_mais_vendas_por_dia = (
    vendas_por_dia_categoria
    .sort_values(["Data de Publicação", "Quantidade Vendida"], ascending=[True, False])
    .drop_duplicates(subset="Data de Publicação")
    .rename(columns={"Categoria do Produto": "Categoria Mais Vendida"})
)

display(
    categoria_mais_vendas_por_dia[
        ["Data de Publicação", "Categoria Mais Vendida", "Quantidade Vendida"]
    ]
    .head(10)
)

# Resumo geral: qual categoria teve mais vendas ao longo dos dias
categoria_mais_vendas_total = (
    vendas_por_dia_categoria.groupby("Categoria do Produto")["Quantidade Vendida"]
    .sum()
    .sort_values(ascending=False)
)

display(categoria_mais_vendas_total)
```

```python
produtos_eletronicos = sorted(
    df1.loc[df1["Categoria do Produto"] == "Electronics", "Produto"].unique()
)

produtos_furniture = sorted(
    df1.loc[df1["Categoria do Produto"] == "Furniture", "Produto"].unique()
)

produtos_office_supplies = sorted(
    df1.loc[df1["Categoria do Produto"] == "Office Supplies", "Produto"].unique()
)

print("Electronics:", produtos_eletronicos)
print("Furniture:", produtos_furniture)
print("Office Supplies:", produtos_office_supplies)
```
